## 📈 Étape 2 : Prétraitement Financier

Conformément au cahier des charges, nous allons :
1. **Nettoyer et synchroniser** les données boursières[cite: 1].
2. **Préparer le décalage temporel (Lag effect)** en vue de nos futurs modèles prédictifs[cite: 1].

💡 **L'innovation de notre modèle : La variable `StockChange`**

Pour mesurer la variation du prix de l'action, nous avons décidé de remplacer le rendement classique par le rendement intra-journalier, défini par l'équation suivante :
$$StockChange = \frac{Close - Open}{Open}$$

**Pourquoi ce choix est-il crucial ?**
Le calcul classique (basé sur la clôture de la veille) inclut ce qu'on appelle l'"Overnight Gap", c'est-à-dire les variations de prix dues aux événements survenus pendant la nuit. En utilisant `StockChange`, nous éliminons toutes les interactions et le bruit générés lorsque la Bourse est fermée. Cette variable nous permet d'isoler de manière pure la réaction des investisseurs *pendant* les heures d'ouverture du marché.

```python
df_finance_top4['Date'] = pd.to_datetime(df_finance_top4['Date']).dt.date
df_finance_top4 = df_finance_top4.sort_values(by=['Ticker', 'Date']).reset_index(drop=True)


df_finance_top4['StockChange_%'] = ((df_finance_top4['Close'] - df_finance_top4['Open']) / df_finance_top4['Open']) * 100


df_finance_top4['StockChange_Lendemain_%'] = df_finance_top4.groupby('Ticker')['StockChange_%'].shift(-1)


df_finance_processed = df_finance_top4.dropna(subset=['StockChange_%', 'StockChange_Lendemain_%']).copy()

chemin_sauvegarde = r"C:\Users\semy4\OneDrive\Bureau\Fintech_project\data\processed\DONNEES_FINANCIERES_PROPRES.csv"
df_finance_processed.to_csv(chemin_sauvegarde, index=False)

print(f"✅ Prétraitement terminé et sauvegardé ! {len(df_finance_processed)} lignes prêtes.")
display(df_finance_processed[['Date', 'Ticker', 'Open', 'Close', 'StockChange_%', 'StockChange_Lendemain_%']].head())
```

## 📊 Étape 3 : Analyse Exploratoire des Données (EDA) Financières

Maintenant que nos données sont propres et que notre variable `StockChange` est calculée, nous passons à l'analyse exploratoire (EDA).

**Objectifs de cette section :**
1. **Étude statistique :** Analyser la moyenne, la variance et la distribution de notre rendement intra-journalier (`StockChange`)[cite: 1].
2. **Visualisation :** Observer l'évolution historique des prix et comparer la volatilité des différentes actions[cite: 1].

```python
sns.set_theme(style="darkgrid")

print(" Lancement de l'Analyse Exploratoire (EDA) Financière...")


print("\n Résumé Statistique du Rendement Intra-journalier (StockChange_%) :")
display(df_finance_processed.groupby('Ticker')['StockChange_%'].describe().round(2))


plt.figure(figsize=(14, 6))
sns.lineplot(data=df_finance_processed, x='Date', y='Close', hue='Ticker', linewidth=1.5)
plt.title(" Évolution des Prix de Clôture (Close)", fontsize=16, fontweight='bold')
plt.xlabel("Date", fontsize=12)
plt.ylabel("Prix (USD)", fontsize=12)
plt.legend(title="Tickers")
plt.tight_layout()
plt.show()


plt.figure(figsize=(14, 6))
sns.kdeplot(data=df_finance_processed, x='StockChange_%', hue='Ticker', fill=True, alpha=0.3)
plt.title(" Volatilité Intra-journalière : Distribution du StockChange (%)", fontsize=16, fontweight='bold')
plt.xlabel("Variation Intra-journalière (%)", fontsize=12)
plt.ylabel("Densité", fontsize=12)
plt.xlim(-10, 10) 
plt.tight_layout()
plt.show()
```

##  Résultats de l'Analyse Exploratoire (EDA)

L'observation de nos graphiques nous permet de comprendre le comportement et le profil de risque de chaque actif financier entre 2020 et 2023. Ces informations sont cruciales pour l'interprétation future de nos modèles de Machine Learning.

### 📈 1. Analyse Temporelle : L'Évolution des Prix de Clôture
Le premier graphique illustre la tendance macroéconomique. Nous observons trois grandes phases : le krach initial du Covid-19 (Mars 2020), suivi d'un marché haussier (Bull Market) spectaculaire porté par les valeurs technologiques, puis une violente correction en 2022 (hausse des taux d'intérêt).

**Comportement par actif :**
*   **🔴 TSLA (Tesla) - Spéculation extrême :** L'action connaît la croissance la plus explosive (passant de moins de 50 USD à plus de 400 USD), mais subit l'effondrement le plus violent fin 2022.
*   **🟢 META (Facebook) - La correction punitive :** Après un pic historique avoisinant les 380 USD, l'action s'effondre en 2022 pour retomber sous la barre des 100 USD.
*   **🔵 AAPL (Apple) - La valeur refuge :** Sa courbe est la plus lisse. La croissance est régulière et l'action résiste beaucoup mieux à la crise de 2022 que ses pairs.
*   **🟠 AMZN (Amazon) - Le cycle complet :** Une trajectoire en cloche classique. L'action monte modérément pour atteindre les 180 USD avant de reperdre la quasi-totalité de ses gains post-Covid.

---

### 🌪️ 2. Analyse de la Volatilité : La Distribution du StockChange (%)
Ce second graphique est fondamental pour notre projet car il mesure le **risque quotidien pur** (la volatilité intra-journalière), en ignorant les variations survenues pendant la nuit lorsque la Bourse est fermée.

En statistiques, la forme de la distribution (la cloche) indique le niveau de risque :
*   **🔵 AAPL (Faible risque) :** La cloche est très haute et fine. La grande majorité de ses journées se clôturent avec des variations minimes (entre -2% et +2%). C'est un actif très rationnel.
*   **🟠 AMZN et 🟢 META (Risque modéré) :** Leurs distributions sont légèrement plus étalées, indiquant une fréquence plus élevée de variations quotidiennes de l'ordre de 3% à 4%.
*   **🔴 TSLA (Risque extrême) :** La courbe est écrasée avec des "queues de distribution" très épaisses qui s'étalent vers -10% et +10%. Il est mathématiquement fréquent que Tesla gagne ou perde plus de 5% au cours d'une seule et même journée.
