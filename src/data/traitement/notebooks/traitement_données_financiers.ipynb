```python
# ============================================================================
#  CONFIGURATION — les deux blocs à adapter
# ============================================================================
PROJET = r"C:\Users\semy4\OneDrive\Bureau\Fintech_project"

import os, warnings
import numpy as np
import pandas as pd

DATA = os.path.join(PROJET, "data", "processed")
DOCS = os.path.join(PROJET, "docs")
os.makedirs(DOCS, exist_ok=True)
pd.set_option("display.width", 170); pd.set_option("display.max_columns", 80)

# --- Bloc 1 : où sont les fichiers ------------------------------------------
MESSAGES   = os.path.join(DATA, "02_StockTwits_SCORED_2020_2022.csv")  
FINANCE    = os.path.join(DATA, "finance_2020_2022_5T.csv")
SORTIE     = os.path.join(DATA, "PANEL_SENTIMENT_WINDOWS_2020_2022_v2.csv")

# --- Bloc 2 : le fuseau des horodatages SOURCE ------------------------------
# StockTwits renvoie de l'UTC. Si tes horodatages sont déjà en heure de New York,
# mets "America/New_York". Si tu n'es pas sûr, le §2 contient un test de diagnostic.
TZ_SOURCE = "UTC"
TZ_MARCHE = "America/New_York"

TICKERS = ["AAPL", "AMZN", "META", "NVDA", "TSLA"]
DEBUT, FIN = "2020-01-02", "2022-03-04"


def charger(chemin):
    ext = os.path.splitext(chemin)[1].lower()
    if ext == ".pkl":     return pd.read_pickle(chemin)
    if ext == ".parquet": return pd.read_parquet(chemin)
    return pd.read_csv(chemin)


msg = charger(MESSAGES)
print(f"{len(msg):,} messages chargés")
print(f"Colonnes disponibles :\n  {list(msg.columns)}")
print(f"\nAperçu :")
print(msg.head(3).to_string())
```
